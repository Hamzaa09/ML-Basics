```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
```

```python
df = sns.load_dataset('diamonds')
df["cut_num"] = df["cut"].map({"Fair": 0, "Good": 1, "Very Good": 2, "Premium": 3, "Ideal": 4})

df.head()
```

```python


X = df[['cut_num', 'carat']]
y = df['price']

model = LinearRegression()
model.fit(X, y)

print(model.predict([[0, 0.5]]))
print(model.predict([[1, 0.5]]))
print(model.predict([[2, 0.5]]))
print(model.predict([[3, 0.5]]))
print(model.predict([[4, 0.5]]))
```
