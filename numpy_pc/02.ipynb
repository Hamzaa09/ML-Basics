```python
# advance indexing

import numpy as np
```

```python
arr = np.arange(0,30, 2)
idx = [1,2,5,7,8,14]

print(arr)
print(arr[idx])
```

```python
matrix = np.reshape(np.arange(1,21), (4,5))

rows = [2, 3, 1]
cols = [0, 4, 3]

print(matrix)
print(matrix[rows, cols])
```

```python
random = np.random.randint(2,100, 10)

mask = random < 18
res = random[mask]

print(mask)
print(res)
print(random[random<18])
```

```python
arr = np.reshape(np.arange(1, 19), (3,6))
print(arr) 

arr[arr < 7] = 1
print(arr)
```

```python
arr = np.reshape(np.arange(1, 19), (3,6))
print(arr)

rows, cols = [0, 1, 2], [2,5,3]
print(arr[np.ix_(rows, cols)])
```

```python
arr = np.arange(20)
res = np.where(arr%2==0, 'hamza', 'odd')

print(res)
```
