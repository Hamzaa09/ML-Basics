```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
```

```python
df = sns.load_dataset("tips")
df.head()
```

```python
sns.lmplot(x="tip", y="total_bill", data=df)
```

```python
X = df[["total_bill"]]
y = df[["tip"]]
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
model.predict([[100]])
```

```python
nf = sns.load_dataset("flights")
nf.head()

sns.lmplot(x="passengers", y="year", data=nf)

X = nf[["passengers"]]
y = nf[["year"]]

model = LinearRegression()
model.fit(X, y)

model.predict([[100]])
```
