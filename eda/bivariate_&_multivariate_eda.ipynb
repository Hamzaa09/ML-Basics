## Categorical vs Categorical

```python
pd.crosstab(titanic['pclass'], titanic['survived'])
```

```python
pd.crosstab(index=[titanic['pclass'], titanic['sex']], columns=titanic['survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['pclass'], titanic['survived']), annot=True, cmap='YlGnBu')
```

```python
sns.heatmap(pd.crosstab(index=[titanic['pclass'], titanic['sex']], columns=titanic['survived']), cmap='YlGnBu', annot=True)
```

```python
sns.countplot(x=titanic['pclass'])
```

```python
sns.countplot(x=titanic['pclass'])
sns.countplot(x='pclass', hue='sex', data=titanic, palette='pastel')
```
