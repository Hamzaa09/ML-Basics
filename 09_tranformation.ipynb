```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
df = np.random.exponential(size=1000, scale=2)
df = pd.DataFrame(df, columns=['Values'])
df.head()
```

```python
sns.histplot(df, kde=True)
```

```python
from sklearn.preprocessing import PowerTransformer
from sklearn.preprocessing import QuantileTransformer

pt_boxcox = PowerTransformer('box-cox', standardize=False)
pt_yeo_jhonson = PowerTransformer('yeo-johnson', standardize=False)
qt_nomal = QuantileTransformer(output_distribution='normal')

df['boxcox'] = pt_boxcox.fit_transform(df[["Values"]] + 1)
df['yeo-jhonson'] = pt_yeo_jhonson.fit_transform(df[["Values"]])
df['quantile'] = qt_nomal.fit_transform(df[["Values"]])

df.head()
```

```python
for col in df.columns:
    sns.histplot(df[col], kde=True)
    plt.show()
```
