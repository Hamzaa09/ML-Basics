```python
import seaborn as sns
import numpy as np
```

```python
data = sns.load_dataset("titanic")
data.head()
```

```python
## check for missing values
data.isnull().sum().sort_values(ascending=False)
```

```python
## drop data with missing values

data.drop('deck', axis=1, inplace=True)
```

```python
data.head()
data.isnull().sum().sort_values(ascending=False)
```

```python
data["age"] = data["age"].fillna(data["age"].median())

data.isnull().sum().sort_values(ascending=False)
```

```python
data["embark_town"] = data["embark_town"].fillna(data["embark_town"].mode()[0])
data["embarked"] = data["embarked"].fillna(data["embarked"].mode()[0])

data.isnull().sum().sort_values(ascending=False)
```

```python
## KNN imputation
from sklearn.impute import KNNImputer
```

```python
imputer = KNNImputer(n_neighbors=5)

data["age"] = imputer.fit_transform(data[["age"]])

data.isnull().sum().sort_values(ascending=False)
```

```python
## Iterative Imputer

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.preprocessing import LabelEncoder

newdata = sns.load_dataset("titanic")
```

```python
newdata.isnull().sum().sort_values(ascending=False)
```

```python
imputer = IterativeImputer(max_iter=5)
label_encoder = {}

cols_to_iter = ["age", "embarked", "embark_town"]

for cols in cols_to_iter:
    le = LabelEncoder()
    data[cols] = le.fit_transform(data[cols])
    label_encoder[cols] = le
    
data.isnull().sum().sort_values()
```

```python
for col in cols_to_iter:
    le = label_encoder[col]
    data[col] = le.inverse_transform(data[col].astype(int))
    
data.head()
```

```python
## Iterative Imputer in one go

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.preprocessing import LabelEncoder

imputer = IterativeImputer(max_iter=10)

columns_to_iter = ["deck", "age", "embarked", "embark_town", "sex", "who", "class", "alive"]
label_encoders = {}

# Encoding
for col in columns_to_iter:
    le = LabelEncoder()
    newdata[col] = le.fit_transform(newdata[col])
    label_encoders[col] = le

# Decoding
for col in columns_to_iter:
    le = label_encoders[col]
    newdata[col] = le.inverse_transform(newdata[col])
    
newdata.head(15)
```
