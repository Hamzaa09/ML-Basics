```python
# Model: DecisionTreeRegressor

from sklearn.tree import DecisionTreeRegressor
```

```python
%%time
df = sns.load_dataset('diamonds')

X = df.drop('price', axis=1)
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Preprocessing
numeric_features = ['carat', 'depth', 'table', 'x', 'y', 'z']
categorical_features = ['cut', 'color', 'clarity']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(), categorical_features)
    ]
)

pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', DecisionTreeRegressor())
])

pipeline.fit(X_train, y_train)

y_predic = pipeline.predict(X_test)


metrics['DTR'].append(mean_squared_error(y_test, y_predic))
metrics['DTR'].append(r2_score(y_test, y_predic))
metrics['DTR'].append(mean_absolute_error(y_test, y_predic))
metrics['DTR'].append(mean_absolute_percentage_error(y_test, y_predic))
metrics['DTR'].append(root_mean_squared_error(y_test, y_predic))


print(f"Mean Squared Error: {mean_squared_error(y_test, y_predic)}")
print(f"R2 Score: {r2_score(y_test, y_predic)}")
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_predic)}")
print(f"Mean Absolute Percentage Error: {mean_absolute_percentage_error(y_test, y_predic)}")
print(f"Root Mean Squared Error: {root_mean_squared_error(y_test, y_predic)}")
```

```python
# Model: RandomForestRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
%%time
df = sns.load_dataset('diamonds')

X = df.drop('price', axis=1)
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Preprocessing
numeric_features = ['carat', 'depth', 'table', 'x', 'y', 'z']
categorical_features = ['cut', 'color', 'clarity']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(), categorical_features)
    ]
)

pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor())
])

pipeline.fit(X_train, y_train)

y_predic = pipeline.predict(X_test)


metrics['RFR'].append(mean_squared_error(y_test, y_predic))
metrics['RFR'].append(r2_score(y_test, y_predic))
metrics['RFR'].append(mean_absolute_error(y_test, y_predic))
metrics['RFR'].append(mean_absolute_percentage_error(y_test, y_predic))
metrics['RFR'].append(root_mean_squared_error(y_test, y_predic))


print(f"Mean Squared Error: {mean_squared_error(y_test, y_predic)}")
print(f"R2 Score: {r2_score(y_test, y_predic)}")
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_predic)}")
print(f"Mean Absolute Percentage Error: {mean_absolute_percentage_error(y_test, y_predic)}")
print(f"Root Mean Squared Error: {root_mean_squared_error(y_test, y_predic)}")
```
